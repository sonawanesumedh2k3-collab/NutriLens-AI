```python
import pandas as pd

df_clean = pd.read_csv("../dataset/packaged_foods_india_cleaned.csv")

print("Dataset loaded successfully!")
print("Shape:", df_clean.shape)
```

```python
nutrition_cols = [
    "calories_kcal",
    "carbohydrates_g",
    "proteins_g",
    "total_fat_g",
    "saturated_fat_g",
    "trans_fat_g",
    "sugar_g",
    "sodium_mg",
    "dietary_fiber_g"
]

per_100g_cols = [
    "calories_per_100g",
    "carbohydrates_per_100g",
    "proteins_per_100g",
    "total_fat_per_100g",
    "saturated_fat_per_100g",
    "trans_fat_per_100g",
    "sugar_per_100g",
    "sodium_per_100g",
    "dietary_fiber_per_100g"
]

for old_col, new_col in zip(nutrition_cols, per_100g_cols):
    df_clean[new_col] = (
        df_clean[old_col] / df_clean["serving_size_g"]
    ) * 100

print("Per-100g columns created successfully!")
```

```python
df_clean[[
    "item_name",
    "serving_size_g",
    "calories_per_100g",
    "sugar_per_100g",
    "sodium_per_100g"
]].head(10)
```

```python
def calorie_points(calories):
    if calories <= 80:
        return 0
    elif calories <= 160:
        return 1
    elif calories <= 240:
        return 2
    elif calories <= 320:
        return 3
    elif calories <= 400:
        return 4
    elif calories <= 480:
        return 5
    elif calories <= 560:
        return 6
    elif calories <= 640:
        return 7
    elif calories <= 720:
        return 8
    elif calories <= 800:
        return 9
    else:
        return 10
```

```python
print("70 kcal  →", calorie_points(70), "points")
print("200 kcal →", calorie_points(200), "points")
print("450 kcal →", calorie_points(450), "points")
print("600 kcal →", calorie_points(600), "points")
print("900 kcal →", calorie_points(900), "points")
```

```python
df_clean["calorie_points"] = df_clean["calories_per_100g"].apply(calorie_points)

print("Calories points created successfully")
```

```python
df_clean[[
    "item_name",
    "calories_per_100g",
    "calorie_points"
]].head(10)
```

```python
def sugar_points(sugar):
    if sugar <= 4.5:
        return 0
    elif sugar <= 9:
        return 1
    elif sugar <= 13.5:
        return 2
    elif sugar <= 18:
        return 3
    elif sugar <= 22.5:
        return 4
    elif sugar <= 27:
        return 5
    elif sugar <= 31.5:
        return 6
    elif sugar <= 36:
        return 7
    elif sugar <= 40.5:
        return 8
    elif sugar <= 45:
        return 9
    else:
        return 10
```

```python
print("3 g sugar   →", sugar_points(3), "points")
print("10 g sugar  →", sugar_points(10), "points")
print("20 g sugar  →", sugar_points(20), "points")
print("30 g sugar  →", sugar_points(30), "points")
print("50 g sugar  →", sugar_points(50), "points")
```

```python
df_clean["sugar_points"] = df_clean["sugar_per_100g"].apply(sugar_points)

print("Sugar points added successfully")
```

```python
df_clean[[
    "item_name",
    "calories_per_100g",
    "calorie_points",
    "sugar_per_100g",
    "sugar_points"
]].head(10)
```

```python
def saturated_fat_points(saturated_fat):
    if saturated_fat <= 1:
        return 0
    elif saturated_fat <= 2:
        return 1
    elif saturated_fat <= 3:
        return 2
    elif saturated_fat <= 4:
        return 3
    elif saturated_fat <= 5:
        return 4
    elif saturated_fat <= 6:
        return 5
    elif saturated_fat <= 7:
        return 6
    elif saturated_fat <= 8:
        return 7
    elif saturated_fat <= 9:
        return 8
    elif saturated_fat <= 10:
        return 9
    else:
        return 10
```

```python
print("0.5 g saturated fat →", saturated_fat_points(0.5), "points")
print("2.5 g saturated fat →", saturated_fat_points(2.5), "points")
print("4.5 g saturated fat →", saturated_fat_points(4.5), "points")
print("7.5 g saturated fat →", saturated_fat_points(7.5), "points")
print("12 g saturated fat  →", saturated_fat_points(12), "points")
```

```python
df_clean["saturated_fat_points"] = (
    df_clean["saturated_fat_per_100g"].apply(saturated_fat_points)
)

print("Saturated fat points created successfully!")
```

```python
def sodium_points(sodium):
    if sodium <= 90:
        return 0
    elif sodium <= 180:
        return 1
    elif sodium <= 270:
        return 2
    elif sodium <= 360:
        return 3
    elif sodium <= 450:
        return 4
    elif sodium <= 540:
        return 5
    elif sodium <= 630:
        return 6
    elif sodium <= 720:
        return 7
    elif sodium <= 810:
        return 8
    elif sodium <= 900:
        return 9
    else:
        return 10
```

```python
print("50 mg sodium   →", sodium_points(50), "points")
print("150 mg sodium  →", sodium_points(150), "points")
print("250 mg sodium  →", sodium_points(250), "points")
print("500 mg sodium  →", sodium_points(500), "points")
print("1000 mg sodium →", sodium_points(1000), "points")
```

```python
def protein_points(protein):
    if protein <= 1.6:
        return 0
    elif protein <= 3.2:
        return 1
    elif protein <= 4.8:
        return 2
    elif protein <= 6.4:
        return 3
    elif protein <= 8.0:
        return 4
    else:
        return 5
```

```python
print("1 g protein   →", protein_points(1), "points")
print("3 g protein   →", protein_points(3), "points")
print("5 g protein   →", protein_points(5), "points")
print("7 g protein   →", protein_points(7), "points")
print("10 g protein  →", protein_points(10), "points")
```

```python
def fiber_points(fiber):
    if fiber <= 0.9:
        return 0
    elif fiber <= 1.8:
        return 1
    elif fiber <= 2.7:
        return 2
    elif fiber <= 3.6:
        return 3
    elif fiber <= 4.5:
        return 4
    else:
        return 5
```

```python
def health_category(score):
    if score <= 5:
        return "Healthy"
    elif score <= 18:
        return "Moderate"
    else:
        return "Unhealthy"
```

```python
df_clean["health_category"] = (
    df_clean["health_score"].apply(health_category)
)
```

```python
features = [
    "calories_per_100g",
    "sugar_per_100g",
    "saturated_fat_per_100g",
    "sodium_per_100g",
    "proteins_per_100g",
    "dietary_fiber_per_100g"
]

target = "health_category"

print("Features:")
print(features)

print("\nTarget:")
print(target)
```

```python
X = df_clean[features]
y = df_clean[target]
```

```python
print(df_clean.shape)
print(features)
print(target)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
# What does this do?
# StandardScaler transforms each feature so that it is approximately on the same scale.
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
# The scaler looks at only the training data and calculates the statistics it needs (mean and standard deviation) for each of our 6 features.
# transform() - It then uses those statistics to convert the training values into scaled values.
```

```python
X_test_scaled = scaler.transform(X_test)
```

```python
print("X_train_scaled shape:", X_train_scaled.shape)
print("X_test_scaled shape :", X_test_scaled.shape)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=1000)
model.fit(X_train_scaled,y_train)
# "Look at the 681 training products and learn how their nutritional values relate to their health categories."
```

```python
y_pred = model.predict(X_test_scaled)
```

```python
print(y_pred[:10])
```

```python
comparison = pd.DataFrame({
    "Actual": y_test.values[:10],
    "Predicted": y_pred[:10]
})

comparison
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test,y_pred)

print(cm)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(7,5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=["Healthy","Moderate","Unhealthy"],
    yticklabels=["Healthy","Moderate","Unhealthy"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - Logistic Regression")

plt.show()
           
```

```python
from sklearn.tree import DecisionTreeClassifier 

# Create the decision tree
tree_model = DecisionTreeClassifier(
    random_state = 42
)
# Train the model
tree_model.fit(X_train_scaled, y_train)
```

```python
tree_pred = tree_model.predict(X_test_scaled)
```

```python
print(tree_pred[:10])
```

```python
tree_accuracy = accuracy_score(y_test,tree_pred)

print("Decision Tree Accuracy:", tree_accuracy)
print("Decision Tree Accuracy (%):", tree_accuracy * 100)
```

```python
# Decision Tree Classification Report
tree_report = classification_report(y_test,tree_pred)

print(tree_report)
```

```python
tree_cm = confusion_matrix(y_test, tree_pred)

print(tree_cm)
```

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    tree_cm,
    annot=True,
    fmt="d",
    xticklabels=["Healthy", "Moderate", "Unhealthy"],
    yticklabels=["Healthy", "Moderate", "Unhealthy"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - Decision Tree")

plt.show()
```
